```python
import json
import re
import unicodedata
from pathlib import Path

import numpy as np
import pandas as pd
import pymupdf
from PIL import Image, ImageOps
from paddleocr import PaddleOCR

DEVICE = "cpu"

# Twice the resolution used when the pages were first rendered for the network
RENDER_DPI = 300

# Detection limit raised from the baseline's 1200 to match the finer render
DET_LIMIT_SIDE_LEN = 1600

# Rotations tried per page, ordered by likelihood. Right angles cover the common
# scanning mistakes; the diagonals are a fallback.
ANGLES = [0, 90, 180, 270, 45]

# The small model rather than tiny: only flagged pages reach this stage, so
# character accuracy now matters more than throughput.
ocr = PaddleOCR(
    ocr_version="PP-OCRv6",
    lang="en",
    text_detection_model_name="PP-OCRv6_small_det",
    text_recognition_model_name="PP-OCRv6_small_rec",
    text_recognition_batch_size=8,
    use_doc_orientation_classify=False,
    use_doc_unwarping=False,
    use_textline_orientation=False,
    device=DEVICE,
)

print("OCR engine ready (PP-OCRv6 small)")
```

```python
def render_page(document_path, page_number, dpi=RENDER_DPI):
    path = Path(document_path)

    if path.suffix.lower() == ".docx":
        from docx2pdf import convert
        temp_dir = Path("data/_temp_render")
        temp_dir.mkdir(parents=True, exist_ok=True)
        convert(str(path), str(temp_dir / f"{path.stem}.pdf"))
        path = temp_dir / f"{path.stem}.pdf"

    with pymupdf.open(path) as document:
        if not 1 <= page_number <= len(document):
            raise ValueError(
                f"Page {page_number} out of range: {path.name} has {len(document)}"
            )
        pixmap = document[page_number - 1].get_pixmap(dpi=dpi)

    return Image.frombytes("RGB", (pixmap.width, pixmap.height), pixmap.samples)


# --- Single-file test --------------------------------------------------------

TEST_DOCUMENT = "data/documents/pdf/passport_normal/passport_normal_00038.pdf"
TEST_PAGE = 2

page_image = render_page(TEST_DOCUMENT, TEST_PAGE)
print(f"Rendered at {RENDER_DPI} DPI: {page_image.size[0]} x {page_image.size[1]} px")

# Look at it before it goes into the engine
page_image.resize((page_image.width // 3, page_image.height // 3))
```

```python
# Convert a PaddleOCR result object into a plain dictionary
def result_payload(result_item):
    if isinstance(result_item, dict):
        payload = result_item
    else:
        payload = getattr(result_item, "json", None)
        if callable(payload):
            payload = payload()

    if isinstance(payload, str):
        payload = json.loads(payload)
    if not isinstance(payload, dict):
        raise TypeError(f"Unexpected OCR result type: {type(result_item)!r}")

    return payload.get("res", payload)


# Run OCR on an image and return all recognised lines as one string
def read_page(image, side_len=DET_LIMIT_SIDE_LEN):
    lines = []
    for result_item in ocr.predict(
        np.asarray(image),
        text_det_limit_side_len=side_len,
        text_det_limit_type="max",
    ):
        payload = result_payload(result_item)
        lines.extend(
            str(text).strip()
            for text in payload.get("rec_texts", [])
            if str(text).strip()
        )
    return "\n".join(lines)


# Rotate an image without cropping it, filling the corners with white
def rotate_page(image, angle):
    image = ImageOps.exif_transpose(image).convert("RGB")
    if angle == 0:
        return image.copy()
    return image.rotate(
        angle,
        resample=Image.Resampling.BICUBIC,
        expand=True,
        fillcolor=(255, 255, 255),
    )


raw_text = read_page(page_image)
print(f"{len(raw_text.splitlines())} lines recognised\n")
print(raw_text)
```

```python
# OCR sometimes returns bracket-like characters instead of the MRZ filler '<'
FILLER_TRANSLATION = str.maketrans({
    "«": "<", "‹": "<", "＜": "<", "〈": "<", "⟨": "<"
})


# Upper-case the text, strip accents, and repair filler characters
def normalize_ocr(value):
    if pd.isna(value):
        return ""
    text = unicodedata.normalize("NFKD", str(value))
    text = "".join(ch for ch in text if not unicodedata.combining(ch))
    text = text.upper().translate(FILLER_TRANSLATION)
    return text.replace("\r\n", "\n").replace("\r", "\n")


# Return lines that could be MRZ: stripped of punctuation and of plausible length
def mrz_candidates(text):
    candidates = []
    for line in text.splitlines():
        compact = re.sub(r"[^A-Z0-9<]", "", line)
        if 30 <= len(compact) <= 50:
            candidates.append(compact)
    return candidates


# TD3 lower line, 44 characters as the ICAO standard defines
MRZ_LOWER = re.compile(
    r"^(?P<passport_no>[A-Z0-9<]{9})"
    r"(?P<passport_check>[0-9<])"
    r"(?P<nationality>[A-Z<]{3})"
    r"(?P<dob>[0-9]{6})"
    r"(?P<dob_check>[0-9<])"
    r"(?P<sex>[MFX<])"
    r"(?P<expiry>[0-9]{6})"
    r"(?P<expiry_check>[0-9<])"
    r"(?P<optional>[A-Z0-9<]{14})"
    r"[0-9<]{2}$"
)

# TD3 upper line: document code, issuing state, then the name field
MRZ_UPPER = re.compile(
    r"^P(?P<doc_type>[A-Z0-9<])"
    r"(?P<issuer>[A-Z<]{3})"
    r"(?P<names>[A-Z<]{39})$"
)


# ICAO check digit: each character times a repeating 7-3-1 weight, modulo 10
def mrz_check_digit(value):
    weights = (7, 3, 1)
    total = 0
    for i, char in enumerate(value):
        if char.isdigit():
            digit = int(char)
        elif char.isalpha():
            digit = ord(char) - ord("A") + 10
        else:
            digit = 0
        total += digit * weights[i % 3]
    return total % 10


# Compare a field against its check digit; None if the digit was not read
def verify_check_digit(value, expected):
    if expected in ("", "<"):
        return None
    return mrz_check_digit(value) == int(expected)


# Convert a six-digit MRZ date (YYMMDD, no century) to ISO format
def parse_mrz_date(value):
    if not value or len(value) != 6 or not value.isdigit():
        return None

    year, month, day = int(value[:2]), int(value[2:4]), int(value[4:6])
    if not (1 <= month <= 12 and 1 <= day <= 31):
        return None

    century = 1900 if year > 30 else 2000
    return f"{century + year:04d}-{month:02d}-{day:02d}"


# Split the MRZ name field into surname and given names
def split_names(names_field):
    if not names_field:
        return None, None
    surname, _, given = names_field.partition("<<")
    clean = lambda part: " ".join(p for p in part.split("<") if p) or None
    return clean(surname), clean(given)


# Pull passport fields from one page of OCR text; None if no MRZ line parses
def extract_fields(text):
    normalised = normalize_ocr(text)
    candidates = mrz_candidates(normalised)

    for line in candidates:
        lower = MRZ_LOWER.match(line)
        if not lower:
            continue

        dob = lower.group("dob")
        expiry = lower.group("expiry")

        fields = {
            "passport_number": lower.group("passport_no").replace("<", "").strip() or None,
            "nationality": lower.group("nationality").replace("<", "").strip() or None,
            "date_of_birth": parse_mrz_date(dob),
            "sex": lower.group("sex") if lower.group("sex") in "MF" else None,
            "expiry_date": parse_mrz_date(expiry),
            "passport_number_valid": verify_check_digit(
                lower.group("passport_no"), lower.group("passport_check")
            ),
            "date_of_birth_valid": verify_check_digit(dob, lower.group("dob_check")),
            "expiry_date_valid": verify_check_digit(expiry, lower.group("expiry_check")),
            "surname": None,
            "given_names": None,
        }

        # The name sits on the upper line, which may not have been read
        for other in candidates:
            upper = MRZ_UPPER.match(other)
            if upper:
                fields["surname"], fields["given_names"] = split_names(upper.group("names"))
                break

        return fields

    return None
```

```python
# Angles tried in turn until the MRZ parses. The engine detects aligned text,
# so a page scanned sideways returns noise at its original orientation.
ANGLES = [0, 90, 180, 270, 45]


# Read one page, trying rotations until the MRZ parses
def extract_from_page(document_path, page_number):
    page_image = render_page(document_path, page_number)

    for angle in ANGLES:
        text = read_page(rotate_page(page_image, angle))
        fields = extract_fields(text)

        if fields is not None:
            # Stops at the first angle that works, so an upright page
            # costs a single OCR pass
            return {**fields, "angle_used": angle, "raw_text": text}

    return None
```

```python
print(f"Document: {Path(TEST_DOCUMENT).name}, page {TEST_PAGE}\n")

result = extract_from_page(TEST_DOCUMENT, TEST_PAGE)

if result is None:
    print("No MRZ line could be parsed")
else:
    print(f"Rotation applied: {result['angle_used']} degrees\n")

    # The MRZ lines the engine managed to read, with their length
    print("MRZ lines read:")
    for line in mrz_candidates(normalize_ocr(result["raw_text"])):
        print(f"  [{len(line):2d}] {line}")

    # The fields parsed out of those lines
    print("\nExtracted fields:")
    for field in ("passport_number", "surname", "given_names",
                  "date_of_birth", "nationality", "sex", "expiry_date"):
        value = result[field]
        if value:
            print(f"  {field:<18} {value}")

    # The standard's own error detection, verifying the values above
    print("\nCheck digits:")
    for field, label in (("passport_number_valid", "passport number"),
                         ("date_of_birth_valid", "date of birth"),
                         ("expiry_date_valid", "expiry date")):
        status = result[field]
        mark = "verified" if status else "failed" if status is False else "not read"
        print(f"  {label:<18} {mark}")
```
