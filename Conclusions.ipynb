<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 12px 0; font-size: 1.2em; font-weight: 700; color: #2c3e50;">מטרת הפרויקט</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 22px 0; line-height: 1.8; color: #2c3e50;">לזהות האם מסמך מכיל תצלום דרכון החושף פרטים אישיים ובאיזה מעמודיו  ולאחר מכן לחלץ ממנו את השדות הרגישים: שם מלא, מספר דרכון, אזרחות, תאריך לידה ותאריך תפוגה. תנאי מרכזי היה שהמערכת תפעל על מחשב רגיל, בלי חומרה ייעודית  וזה מה שהכתיב את רוב ההחלטות בדרך.</div>

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 12px 0; font-size: 1.2em; font-weight: 700; color: #2c3e50;">מהלך העבודה</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 14px 0; line-height: 1.8; color: #2c3e50;"><b>זיהוי לפי כללים:</b> מנוע OCR קורא את הטקסט שבתמונה, וביטויים רגולריים מחפשים בו סימנים שמאפיינים דרכון — בעיקר את אזור ה-MRZ, שהמבנה שלו קבוע בתקן. בנינו מאגר מסמכים שבתוכם הוטמעו תצלומי דרכון, חלקם תקינים וחלקם משובשים, כדי לראות איפה הגישה נשברת.</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 14px 0; line-height: 1.8; color: #2c3e50;"><b>זיהוי חזותי:</b> התוצאות הובילו אותנו לרשת CNN. היא מסווגת עמודים לפי המראה שלהם ולא לפי הטקסט שבהם, ולכן לא תלויה בכך שה-OCR יקרא נכון.</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 22px 0; line-height: 1.8; color: #2c3e50;"><b>חילוץ הפרטים:</b> לאחר שהזיהוי מסתמך על הרשת, מורץ החילוץ על העמודים שסומנו בלבד: העמוד מרונדר ברזולוציה גבוהה, מנוע OCR חזק קורא אותו, ושורת ה-MRZ מפורקת לשדות. בשלב זה בוצעה הדגמה על מסמך אחד, כדי להראות שהתהליך עובד מקצה לקצה.</div>

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 12px 0; font-size: 1.2em; font-weight: 700; color: #2c3e50;">מה למדנו</div>
<div style="direction: rtl !important; text-align: right !important; background: #f7fafc; border-right: 4px solid #3498db; border-radius: 8px; padding: 14px 18px; margin-bottom: 10px;">
<div style="direction: rtl !important; text-align: right !important; font-weight: 700; color: #2471a3; margin-bottom: 5px;">הזיהוי לפי כללים מדויק, אבל מוצא מעט מדי</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50;">זוהו 2 מתוך 31 עמודי דרכון בלבד — אבל אף עמוד לא סווג בטעות. כל זיהוי שהתקבל היה נכון, ומכאן שה-OCR לא ממציא טקסט שלא קיים בתמונה.</div>
</div>
<div style="direction: rtl !important; text-align: right !important; background: #f7fafc; border-right: 4px solid #3498db; border-radius: 8px; padding: 14px 18px; margin-bottom: 10px;">
<div style="direction: rtl !important; text-align: right !important; font-weight: 700; color: #2471a3; margin-bottom: 5px;">הבעיה לא הייתה בקריאה אלא בתבנית</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50;">שורות MRZ שנקראו חלקית נדחו, כי התקן דורש 44 תווים בדיוק. אותן שורות נקראו במלואן בשלב החילוץ, עם מנוע חזק יותר וברזולוציה גבוהה יותר — כלומר המידע היה בתמונה כל הזמן.</div>
</div>
<div style="direction: rtl !important; text-align: right !important; background: #f7fafc; border-right: 4px solid #3498db; border-radius: 8px; padding: 14px 18px; margin-bottom: 10px;">
<div style="direction: rtl !important; text-align: right !important; font-weight: 700; color: #2471a3; margin-bottom: 5px;">לרכך את התבנית לא היה עוזר</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50;">סביב התצלום יש הרבה טקסט, שעלול להכיל מחרוזות דומות לאלה שאנחנו מחפשים. תבנית מתירנית הייתה מוצאת יותר, אבל גם טועה יותר — ולכן הפתרון היה צריך לבוא ממקום אחר.</div>
</div>
<div style="direction: rtl !important; text-align: right !important; background: #f7fafc; border-right: 4px solid #3498db; border-radius: 8px; padding: 14px 18px; margin-bottom: 22px;">
<div style="direction: rtl !important; text-align: right !important; font-weight: 700; color: #2471a3; margin-bottom: 5px;">הזיהוי החזותי הפוך בדיוק — מוצא הרבה, וטועה לפעמים</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50;">הרשת איתרה 30 מתוך 34 עמודי הדרכון בקבוצת המבחן (Recall של 88%), במחיר של 14 עמודים שסומנו בטעות (Precision של 68%). זו תמונת ראי של הגישה הקודמת, וזה מתאים לתפקיד: הרשת משמשת כמסננת, ועמוד שסומן בטעות רק מוסיף עבודה לשלב החילוץ, בעוד שעמוד שפוספס אבוד לחלוטין. סף ההחלטה נבחר בהתאם, כך שיעדיף רגישות על פני דיוק.</div>
</div>

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 12px 0; font-size: 1.2em; font-weight: 700; color: #2c3e50;">המסקנה</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 22px 0; line-height: 1.8; color: #2c3e50;">שתי הגישות נכשלות במקומות הפוכים: הכללים מדויקים אך מוצאים מעט, והרשת מוצאת הרבה אך מסמנת גם עמודים שאינם דרכון. השילוב ביניהן הוא מה שעובד  הרשת מסננת ומשאירה מעט עמודים, ואז אפשר להרשות מנוע OCR כבד ומדויק שיקרא אותם. בזה גם התנאי המקורי נשמר, כי הכל עדיין רץ על מחשב רגיל.</div>

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 12px 0; font-size: 1.2em; font-weight: 700; color: #2c3e50;">מגבלות וכיווני המשך</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0; line-height: 1.8; color: #2c3e50;">המאגר בנוי מתצלומים מלאכותיים בלבד, ולכן שווה לבחון את המערכת גם על מסמכים שצולמו במציאות. החילוץ הודגם על מסמך אחד, וכדי לאמוד את שיעור ההצלחה שלו נדרשת הרצה על מדגם רחב יותר. כמו כן, כל רכיב נבדק בנפרד והביצועים של השרשרת המלאה  סיווג ואחריו חילוץ  עוד לא נמדדו.</div>
