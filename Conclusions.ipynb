<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 12px 0; font-size: 1.2em; font-weight: 700; color: #2c3e50;">מטרת הפרויקט</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 22px 0; line-height: 1.8; color: #2c3e50;">לזהות האם מסמך מכיל תצלום דרכון ובאיזה מעמודיו, ובהמשך לחלץ ממנו את השדות הרגישים. האילוץ המרכזי היה שהמערכת תפעל על מעבד סטנדרטי בלבד, והוא שהכתיב את רוב ההחלטות לאורך הדרך.</div>

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 12px 0; font-size: 1.2em; font-weight: 700; color: #2c3e50;">מהלך העבודה</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 14px 0; line-height: 1.8; color: #2c3e50;"><b>זיהוי לפי כללים:</b> OCR קורא את הטקסט, וביטויים רגולריים מחפשים בו את אזור ה-MRZ. נבנה מאגר מסמכים עם דרכונים תקינים ומשובשים, כדי לראות היכן הגישה נשברת.</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 14px 0; line-height: 1.8; color: #2c3e50;"><b>זיהוי חזותי:</b> המאגר הורחב, מכל עמוד חולץ ייצוג וקטורי באמצעות רשת מאומנת מראש, ועל גביו אומנו מודל בסיס פשוט ורשת CNN. הסיווג נעשה לפי מראה העמוד ולא לפי הטקסט שבו.</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 22px 0; line-height: 1.8; color: #2c3e50;"><b>חילוץ הפרטים:</b> על העמוד שזוהה בלבד מורץ מנוע OCR חזק ברזולוציה גבוהה, ושורת ה-MRZ מפורקת לשדותיה. התהליך הודגם על מסמך אחד.</div>

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 12px 0; font-size: 1.2em; font-weight: 700; color: #2c3e50;">מה למדנו</div>

<div style="direction: rtl !important; text-align: right !important; font-weight: 700; color: #2471a3; margin-bottom: 5px;">שתי הגישות נכשלות במקומות הפוכים</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50;">הכללים זיהו 2 מתוך 31 עמודי דרכון בלבד, אך לא טעו אף פעם. ה-CNN איתר את כל 36 עמודי הדרכון בקבוצת המבחן, במחיר של 9 עמודים שסומנו בטעות. חלוקה זו מתאימה לתפקיד, שכן עמוד שסומן בטעות רק מוסיף עבודה, ואילו עמוד שפוספס אבוד.</div>

<div style="direction: rtl !important; text-align: right !important; font-weight: 700; color: #2471a3; margin-bottom: 5px;">תוצאה מושלמת על סט המבחן אינה הוכחה</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50;">מודל הבסיס סיווג נכון את כל סט המבחן שלו, אך על מאגר בלתי תלוי פספס 3 מתוך 31 עמודי דרכון, בעוד שה-CNN סיווג נכון את כל 141. בחינת דוגמאות בודדות הראתה שמודל הבסיס נשען על צבע ועל מיקום בראש העמוד, ואילו ה-CNN יציב בציוניו ללא תלות בשניהם.</div>

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 12px 0; font-size: 1.2em; font-weight: 700; color: #2c3e50;">המסקנה</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 22px 0; line-height: 1.8; color: #2c3e50;">האילוץ שהמערכת תפעל על מעבד סטנדרטי הוא שהגדיר את הבעיה: כשצריך לעבד את כל העמודים, אין ברירה אלא מנוע קטן ורזולוציה נמוכה, ובתנאים אלה הגישה מבוססת הכללים נכשלה. הזיהוי החזותי מצמצם את היקף העיבוד למעט עמודים, ובכך מאפשר להריץ עליהם מנוע כבד ומדויק שנפסל קודם לכן. מה שפתר את הבעיה אינו אלגוריתם אחר אלא סדר אחר.</div>

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 12px 0; font-size: 1.2em; font-weight: 700; color: #2c3e50;">מגבלות וכיווני המשך</div>
