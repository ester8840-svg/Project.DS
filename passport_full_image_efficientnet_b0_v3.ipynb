```python
from pathlib import Path
import random

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from PIL import Image
from tqdm.auto import tqdm

import torch
import torch.nn as nn
from torchvision.models import EfficientNet_B0_Weights, efficientnet_b0

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    ConfusionMatrixDisplay,
    classification_report,
    roc_auc_score,
)
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

DATA_DIR = Path("data/pages")
OUTPUT_DIR = Path("passport_full_image_output")
LABELS_FILE = Path("passport_page_labels_big_dataset.csv")
DEVICE = torch.device("cpu")

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

labels_df = pd.read_csv(LABELS_FILE, encoding="utf-8-sig")

files_df = pd.DataFrame({
    "image_name": labels_df["image_path"].map(lambda path: Path(path).name),
    "image_path": labels_df["image_path"]
})
```

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; background: #f7fafc; border: 1px solid #e1e8ed; border-right: 5px solid #3498db; border-radius: 12px; padding: 20px 24px;">
<div style="direction: rtl !important; text-align: right !important; font-weight: 700; color: #2471a3; margin-bottom: 6px;">טעינת קובץ תמונה</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50;">תמונות מסוימות עשויות להיות בגוני אפור או עם שקיפות מסויימת, בעוד EfficientNet מצפה לקלט בעל שלושה ערוצים. לכן נעביר את הקלט לפורמט צבע אחיד מסוג RGB.</div>
</div>

```python
def open_rgb(path):
    return Image.open(path).convert("RGB")

sample_path = files_df.iloc[0]["image_path"]
sample_image = open_rgb(sample_path)

plt.figure(figsize=(6, 6))
plt.imshow(sample_image)
plt.title(f"Original: {sample_image.size}")
plt.axis("off")
plt.show()
```

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; background: #f7fafc; border: 1px solid #e1e8ed; border-right: 5px solid #3498db; border-radius: 12px; padding: 20px 24px;">
<div style="direction: rtl !important; text-align: right !important; font-weight: 700; color: #2471a3; margin-bottom: 6px;">התאמת גודל</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50;">דף A4 סטנדרתי הוא לא ריבוע מושלם ולכן כדי שבשלבים הבאים לא נחתוך פרטים                                                                 חשובים נוסיף לו שוליים                                                                                       לבנים כדי להפוך את התמונה לריבוע. אם נשנה ישירות תמונת A4 מלבנית ל־224×224 כפי שדורש המודל בו בחרנו התמונה תימתח                              לרוחב ולכן הטקסט והצורות הגאומטיריות בדרכון יעבור עיוות.</div>
</div>

```python
def letterbox_square(image, fill=(255, 255, 255)):
    side = max(image.size)
    canvas = Image.new("RGB", (side, side), fill)
    x = (side - image.width) // 2
    y = (side - image.height) // 2
    canvas.paste(image, (x, y))
    return canvas

sample_square = letterbox_square(sample_image)

plt.figure(figsize=(6, 6))
plt.imshow(sample_square)
plt.title(f"Square with padding: {sample_square.size}")
plt.axis("off")
plt.show()
```

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; background: #f7fafc; border: 1px solid #e1e8ed; border-right: 5px solid #3498db; border-radius: 12px; padding: 20px 24px;">
<div style="direction: rtl !important; text-align: right !important; font-weight: 700; color: #2471a3; margin-bottom: 6px;">הורדת ממדים</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50;">בשלב זה נטען EfficientNet-B0 עם המשקלים המאומנים מראש והעיבוד המקדים המתאים להם. תחילה התמונה עוברת הקטנה, כלומר הפחתת מספר הפיקסלים שלה תוך שמירה ככל האפשר על התוכן החזותי. לדוגמה, תמונה בגודל 1650×1650 מוקטנת ל-256×256, ולכן מספר רב של פיקסלים מקוריים מיוצג על ידי פיקסל חדש אחד, שערך הצבע שלו מחושב מהפיקסלים הסמוכים בתמונה המקורית. התמונה נשארת דומה מבחינה חזותית, אך מאבדת פרטים קטנים כמו אותיות דקות, קווים עדינים ואזור ה-MRZ.</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50; margin-top: 10px;">בשלב השני נלקח מהמרכז אזור ריבועי בגודל 224×224 פיקסלים, שהוא גודל הקלט שעליו אומנו משקלי הרשת כלומר נחתכים 16 פיקסלים מכל צד. החיתוך עלול להסיר מידע מקצוות העמוד, אך הריפוד לריבוע שקדם לו מבטיח שהתוכן עצמו נשאר במרכז.</div>
</div>

```python
weights = EfficientNet_B0_Weights.DEFAULT
preprocess = weights.transforms()
encoder = efficientnet_b0(weights=weights)
encoder.classifier = nn.Identity()
encoder = encoder.to(DEVICE).eval()
```

```python
image_tensor = preprocess(sample_square)

# Reverse ImageNet normalization for display
mean = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
std = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)

resized_image = image_tensor * std + mean
resized_image = resized_image.clamp(0, 1)
resized_image = resized_image.permute(1, 2, 0).numpy()

plt.figure(figsize=(6, 6))
plt.imshow(resized_image)
plt.title(
    f"EfficientNet input: "
    f"{image_tensor.shape[2]}×{image_tensor.shape[1]}"
)
plt.axis("off")
plt.show()
```

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; background: #f7fafc; border: 1px solid #e1e8ed; border-right: 5px solid #3498db; border-radius: 12px; padding: 20px 24px;">
<div style="direction: rtl !important; text-align: right !important; font-weight: 700; color: #2471a3; margin-bottom: 6px;">מעבר לוקטור</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50;">בשלב זה נהפוך את התמונה מאובייקט תמונה למערך מספרי רב־ממדי שבו כל מספר מייצג ערך צבע של פיקסל. לאחר ההמרה, תמונת RGB בגודל 224×224 מקבלת מבנה של מערך (3,224,224), כאשר 3 הם ערוצי האדום, הירוק והכחול. מבנה מספרי זה מאפשר ל־EfficientNet לבצע על התמונה פעולות מתמטיות כמו קונבולוציות וחילוץ מאפיינים.</div>
</div>

```python
image_tensor = image_tensor.unsqueeze(0)

with torch.inference_mode():
    encoder_result = encoder(image_tensor)

vector = encoder_result.squeeze(0).numpy()

print("Encoder result shape:", encoder_result.shape)
print("Vector shape:", vector.shape)
print("First 10 values:")
print(vector[:10])
```

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; background: #f7fafc; border: 1px solid #e1e8ed; border-right: 5px solid #3498db; border-radius: 12px; padding: 20px 24px;">
<div style="direction: rtl !important; text-align: right !important; font-weight: 700; color: #2471a3; margin-bottom: 6px;">המרת סט הנתונים לוקטורים</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50;">עכשיו נבצע את ההמרה שעד כו הצגנו לדף בודד לכלל סט האימון הסינטתי המורחב שיצרנו קודם.</div>
</div>

```python
vectors = []
metadata = []

for row in tqdm(
    files_df.itertuples(index=False),
    total=len(files_df),
    desc="Images"
):
    image = open_rgb(row.image_path)
    square = letterbox_square(image)

    # prepares the square image for EfficientNet-B0
    image_tensor = preprocess(square)

    # adding the batch dimension so EfficientNet can process it
    image_tensor = image_tensor.unsqueeze(0)

    # disables gradient tracking during vector extraction,
    # reducing memory usage and improving inference speed.
    with torch.inference_mode():
        vector = encoder(image_tensor)

    vector = vector.squeeze(0).numpy().astype(np.float32)
    vectors.append(vector)

    metadata.append({
        "image_name": row.image_name,
        "image_path": row.image_path,
        "vector_row": len(vectors) - 1,
    })

X = np.stack(vectors).astype(np.float32)
metadata_df = pd.DataFrame(metadata)

print("Embedding matrix:", X.shape)
print("First vector shape:", X[0].shape)

np.savez_compressed(
    OUTPUT_DIR / "full_page_embeddings.npz",
    X=X,
    image_name=metadata_df["image_name"].to_numpy(dtype=str),
)

metadata_df.to_csv(
    OUTPUT_DIR / "full_page_embedding_metadata.csv",
    index=False,
    encoding="utf-8-sig",
)

print("Saved vectors:", OUTPUT_DIR / "full_page_embeddings.npz")
print("Saved metadata:", OUTPUT_DIR / "full_page_embedding_metadata.csv")
```

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; background: #f7fafc; border: 1px solid #e1e8ed; border-right: 5px solid #3498db; border-radius: 12px; padding: 20px 24px;">
<div style="direction: rtl !important; text-align: right !important; font-weight: 700; color: #2471a3; margin-bottom: 6px;">בדיקת איכות הווקטורים</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50;">כדי לבדוק את איכות הווקטורים אנחנו מאמנים מסווג בסיסי על וקטורי 1,280 המאפיינים שהופקו מכל תמונה, ובוחנים האם הייצוג שיצרה EfficientNet מאפשר להפריד בין עמודים עם דרכון לעמודים ללא דרכון.</div>
<div style="direction: rtl !important; text-align: right !important; line-height: 1.75; color: #2c3e50; margin-top: 10px;">בחרנו ברגרסיה לוגיסטית משום שהיא מסווג פשוט ומהיר, המתאים כמודל בסיס לווקטורים שהופקו מרשת מאומנת מראש. היא מפרידה באמצעות גבול החלטה ליניארי בלבד, ולכן אם היא מצליחה סימן שהווקטורים עצמם מפרידים היטב, ולא נדרשת מורכבות נוספת שעלולה לגרום להתאמת יתר.</div>
</div>

```python
model_df = metadata_df[["image_path", "vector_row"]].merge(
    labels_df,
    on="image_path",
    how="inner",
    validate="one_to_one",
)

vector_rows = model_df["vector_row"].astype(int).to_numpy()
X_model = X[vector_rows]
y = model_df["label"].astype(int).to_numpy()


all_indices = np.arange(len(model_df))
train_idx, test_idx = train_test_split(
    all_indices,
    test_size=0.20,
    random_state=SEED,
    stratify=y,
)

classifier = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=SEED,
    ),
)

classifier.fit(X_model[train_idx], y[train_idx])
predictions = classifier.predict(X_model[test_idx])
probabilities = classifier.predict_proba(X_model[test_idx])[:, 1]

print(classification_report(
    y[test_idx],
    predictions,
    target_names=["No passport", "Passport"],
    digits=4,
    zero_division=0,
))

if len(np.unique(y[test_idx])) == 2:
    print("ROC AUC:", round(roc_auc_score(y[test_idx], probabilities), 4))

ConfusionMatrixDisplay.from_predictions(
    y[test_idx],
    predictions,
    display_labels=["No passport", "Passport"],
)
plt.show()

joblib.dump(
    classifier,
    OUTPUT_DIR / "full_page_logistic_baseline.joblib",
)
```
