<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 18px 0; line-height: 1.8; color: #2c3e50; font-size: 1.05em;">ארגונים רבים נדרשים לעבד מסמכים מזהים המכילים <b>מידע אישי ורגיש</b>, ובחלק ניכר מהמקרים קליטת הנתונים ואימות הזהות עדיין מתבצעים ידנית. תהליך זה איטי, מועד לטעויות, וחושף מידע פרטי לעיניים אנושיות שלא לצורך.</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 18px 0; line-height: 1.8; color: #2c3e50; font-size: 1.05em;">בנוסף, ריבוי המסמכים העוברים בין ארגונים חושף אותם לאתגרים בשני מישורים: באבטחת המידע, הנפגעת עקב דליפת מסמכים רגישים; וברגולציה, המחייבת פעמים רבות מערכות כבדות ויקרות כדי לעמוד בתקן הנדרש.</div>

<h3 style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 18px 0; font-size: 1.5em; font-weight: 700; color: #2c3e50;">אילוץ תכנוני: פתרון מקומי מבוסס CPU</h3>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 18px 0; line-height: 1.8; color: #2c3e50; font-size: 1.05em;">מתוך שני החסמים שתוארו לעיל נגזר אילוץ תכנוני מרכזי: המערכת תפעל באופן מקומי בלבד, ועל גבי מעבד מחשב סטנדרטי, ללא תלות בשירותי ענן חיצוניים ובלא חומרה ייעודית. אילוץ זה נועד להבטיח שהפתרון יישאר זול ופשוט להטמעה.</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0; line-height: 1.8; color: #2c3e50; font-size: 1.05em;">אילוץ זה מתנגש עם דרישת העמידות: ככל שהמערכת נדרשת להתמודד עם אתגרי איכות וצורת סריקת הדרכון, כך גדלה הנטייה לפתרונות עתירי חישוב. התנגשות זו מלווה את הפרויקט לכל אורכו, והיא שמכתיבה את סדר הבחינה של הגישות השונות.</div>

<h2 style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0; font-size: 1.9em; font-weight: 800; color: #ffffff;">שלב ראשוני — הערכת קו בסיס</h2>

<div style="direction: rtl !important; text-align: right !important; font-family: inherit; background: #ffffff; border: 1px solid #e1e8ed; border-radius: 12px; padding: 36px 34px;">

<h3 style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 18px 0; font-size: 1.5em; font-weight: 700; color: #2c3e50;">הגישה הנבחנת בשלב זה: מודל מבוסס OCR וחוקיות (Regex)</h3>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 18px 0; line-height: 1.8; color: #2c3e50; font-size: 1.05em;">הגישה הראשונה שנבחנת מתבססת על תהליך דו-שלבי: תחילה, הפעלת מנוע זיהוי תווים אופטי (OCR) לחילוץ הטקסט מהתמונה, ולאחר מכן יישום של ביטויים רגולריים (Regex) לאיתור תבניות טקסט ייחודיות המאפיינות דרכונים בראשן אזור ה-MRZ, המוגדר תחת תקן בינלאומי.</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0; line-height: 1.8; color: #2c3e50; font-size: 1.05em;">הבחירה להתחיל בגישה זו נובעת מהיעילות שלה: היא אינה דורשת אימון מראש ומאפשרת ריצה חלקה על מעבד סטנדרטי , מה שעונה על אילוצי הפרויקט. בנוסף, היא משמשת עבורנו כקו בסיס  נקודת ייחוס ברורה לביצועי המערכת. לאחר בחינת תוצאות שלב זה וקבלת אינדיקציה היכן גישה זו נכשלת, נוכל להתקדם בצורה מבוססת לגישה מורכבת יותר.</div>

<h3 style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 18px 0; font-size: 1.5em; font-weight: 700; color: #2c3e50;">שלב ראשוני — הערכת קו בסיס (Baseline)</h3>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 16px 0; line-height: 1.8; color: #2c3e50; font-size: 1.05em;">לצורך בחינת גבולות הגישה והערכת קו הבסיס, בנינו מאגר מסמכים סינתטי בפורמטים נפוצים (Word ו-PDF). מאגר זה תוכנן במטרה לדמות תרחישי מציאות מגוונים: הוא מבוסס על תצלומי דרכונים פיקטיביים ממאגר נתונים פתוח (HuggingFace), כאשר חלקם שולבו במסמכים בתצוגה ברורה ותקנית, ואילו חלקם שובשו במכוון באמצעות טשטוש, שינוי זווית, רעש או חיתוך. מניפולציות אלו נועדו לייצר מקרי קצה שמאתגרים את המערכת ומשמשים כמבחן מאמץ לאלגוריתם הזיהוי.</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0 0 16px 0; line-height: 1.8; color: #2c3e50; font-size: 1.05em;">כדי להתמודד עם האתגר שמציבים הפורמטים השונים של המסמכים, בחרנו לנרמל אותם תחילה לתמונות, כאשר כל דף במסמך מהווה תמונה. תהליך המרה זה הוא שלב הכרחי שמבטיח כי כל שלבי העיבוד הבאים יפעלו על קלט חזותי אחיד וסטנדרטי, ללא קשר לפורמט הקובץ המקורי.</div>
<div style="direction: rtl !important; text-align: right !important; font-family: inherit; margin: 0; line-height: 1.8; color: #2c3e50; font-size: 1.05em;">על גבי התמונות המנורמלות מופעל מנוע ה-OCR (Optical Character Recognition – זיהוי תווים אופטי). טכנולוגיה זו מנתחת את הפיקסלים שבתמונה, מזהה בהם צורות של אותיות ומספרים, ומתרגמת אותם למחרוזות של טקסט דיגיטלי חי. זהו למעשה הגישור ההכרחי בין עולם הנתונים הוויזואליים לבין שלב הזיהוי (באמצעות Regex), אשר דורש טקסט קריא כדי לאתר את התבניות ולהכריע האם קיים דרכון בעמוד.</div>
</div>
